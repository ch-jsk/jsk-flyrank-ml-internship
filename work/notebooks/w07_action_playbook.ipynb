## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

The goal of this playbook is to identify pages that appear to underperform relative to their visibility.

Each recommendation consists of:

- a priority score
- a reason code
- a suggested action
- a confidence note

The output is intended as decision-support rather than an automated publishing system.

```python
import pandas as pd

from sklearn.model_selection import train_test_split

df = pd.read_csv(
    "data/raw/content_refresh_anonymized.csv"
)
```

```python
import pandas as pd
import numpy as np

queue = df.copy()

queue["priority_score"] = (
    queue["impressions_90d"] * 0.4
    + (10 - queue["avg_position"]) * 100
)

conditions = [
    (queue["ctr"] < 0.5),
    (queue["avg_position"] < 5),
    (queue["engagement_rate"] < 50)
]

choices = [
    "Low CTR",
    "Strong ranking",
    "Weak engagement"
]

queue["reason_code"] = np.select(
    conditions,
    choices,
    default="Monitor"
)

actions = {
    "Low CTR": "Rewrite title and meta description",
    "Strong ranking": "Improve content quality",
    "Weak engagement": "Improve page structure",
    "Monitor": "Observe performance"
}

queue["action"] = queue["reason_code"].map(actions)

queue = queue.sort_values(
    "priority_score",
    ascending=False
)

queue[
    [
        "content_id",
        "priority_score",
        "reason_code",
        "action"
    ]
].head(20)
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

**Human review**

Every recommendation should be reviewed manually before any changes are made.

Reviewers should verify:

- search intent alignment;
- content quality;
- seasonality effects;
- recent updates;
- competitive landscape.

**No-go list**

The following actions should never be fully automated:

- publishing content;
- deleting content;
- changing content strategy;
- making large-scale SEO decisions;
- interpreting causation from correlations.
