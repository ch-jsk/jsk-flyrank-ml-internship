## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule such as "review every page with CTR below 1%" would ignore important context like search position, impressions, freshness, content type, and engagement.

Pages naturally receive different CTRs depending on where they rank in search results. A ranking or scoring approach can combine multiple signals to prioritise pages more effectively than a single threshold-based rule.

The purpose of the model is to support human decision-making by identifying pages that deserve review first.
