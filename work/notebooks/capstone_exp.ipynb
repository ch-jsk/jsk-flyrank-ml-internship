```python
# March 2026 development table — DuckDB aggregation
# June 2026 is not accessed by this query.

con.execute(f"""
CREATE OR REPLACE TABLE march_dev AS
WITH monthly AS (
    SELECT
        client_hash_id,
        content_hash_id,

        COUNT(DISTINCT report_date) AS calendar_days_observed,

        COUNT(DISTINCT report_date)
            FILTER (WHERE gsc_data_available IS TRUE)
            AS gsc_days_available,

        COUNT(DISTINCT report_date)
            FILTER (WHERE ga4_data_available IS TRUE)
            AS ga4_days_available,

        SUM(gsc_impressions)
            FILTER (WHERE gsc_data_available IS TRUE)
            AS impressions_march,

        SUM(gsc_clicks)
            FILTER (WHERE gsc_data_available IS TRUE)
            AS clicks_march,

        SUM(gsc_sum_position)
            FILTER (WHERE gsc_data_available IS TRUE)
            AS position_sum_march,

        SUM(gsc_impressions)
            FILTER (
                WHERE gsc_data_available IS TRUE
                  AND gsc_avg_position IS NOT NULL
            ) AS impressions_with_position,

        SUM(ga4_sessions)
            FILTER (WHERE ga4_data_available IS TRUE)
            AS sessions_march,

        SUM(ga4_engaged_sessions)
            FILTER (WHERE ga4_data_available IS TRUE)
            AS engaged_sessions_march,

        SUM(ga4_pageviews)
            FILTER (WHERE ga4_data_available IS TRUE)
            AS pageviews_march,

        SUM(scroll_events)
            FILTER (WHERE ga4_data_available IS TRUE)
            AS scroll_events_march

    FROM read_parquet('{march_path}')
    WHERE report_date >= DATE '2026-03-01'
      AND report_date <  DATE '2026-04-01'
    GROUP BY client_hash_id, content_hash_id
)
SELECT
    m.*,

    100.0 * clicks_march
        / NULLIF(impressions_march, 0) AS ctr_march_pct,

    position_sum_march::DOUBLE
        / NULLIF(impressions_with_position, 0)
        AS weighted_position_march,

    100.0 * engaged_sessions_march
        / NULLIF(sessions_march, 0)
        AS engagement_rate_march_pct,

    100.0 * scroll_events_march
        / NULLIF(pageviews_march, 0)
        AS scroll_events_per_100_pageviews,

    c.content_type,
    c.content_created_date,
    c.content_updated_date,
    c.search_volume,
    c.competition,
    c.main_intent,
    c.word_count,

    cl.gsc_data_start,
    cl.ga4_data_start

FROM monthly m
LEFT JOIN read_parquet('{content_path}') c
    ON m.client_hash_id = c.client_hash_id
   AND m.content_hash_id = c.content_hash_id
LEFT JOIN read_parquet('{clients_path}') cl
    ON m.client_hash_id = cl.client_hash_id
""")

print("March development table created.")
```

```python
# Aggregate checks only; do not display individual page identifiers.

march_summary = con.execute("""
    SELECT
        COUNT(*) AS client_content_pairs,
        COUNT(DISTINCT client_hash_id) AS clients,
        COUNT(DISTINCT content_hash_id) AS content_items,

        COUNT(*) FILTER (
            WHERE content_type IS NULL
        ) AS unmatched_content_dimension,

        COUNT(*) FILTER (
            WHERE gsc_data_start IS NULL
        ) AS unmatched_client_dimension,

        COUNT(*) FILTER (
            WHERE impressions_march > 0
        ) AS pairs_with_search_impressions,

        COUNT(*) FILTER (
            WHERE impressions_march > 0
              AND ctr_march_pct IS NOT NULL
        ) AS pairs_with_defined_ctr,

        COUNT(*) FILTER (
            WHERE weighted_position_march IS NOT NULL
        ) AS pairs_with_position,

        COUNT(*) FILTER (
            WHERE sessions_march IS NOT NULL
        ) AS pairs_with_ga4_sessions

    FROM march_dev
""").fetchdf()

display(march_summary)

print("\nCTR distribution (percentage points):")

display(con.execute("""
    SELECT
        COUNT(*) AS n,
        ROUND(AVG(ctr_march_pct), 4) AS mean_ctr_pct,
        ROUND(MEDIAN(ctr_march_pct), 4) AS median_ctr_pct,
        ROUND(QUANTILE_CONT(ctr_march_pct, 0.75), 4) AS p75_ctr_pct,
        ROUND(AVG(weighted_position_march), 3) AS mean_position
    FROM march_dev
    WHERE impressions_march > 0
      AND ctr_march_pct IS NOT NULL
""").fetchdf())
```

```python
client_join_audit = con.execute(f"""
WITH march_clients AS (
    SELECT DISTINCT client_hash_id
    FROM read_parquet('{march_path}')
),
known_clients AS (
    SELECT client_hash_id
    FROM read_parquet('{clients_path}')
)
SELECT
    COUNT(*) AS distinct_march_clients,
    COUNT(*) FILTER (
        WHERE k.client_hash_id IS NOT NULL
    ) AS clients_with_dimension_match,
    COUNT(*) FILTER (
        WHERE k.client_hash_id IS NULL
    ) AS clients_without_dimension_match
FROM march_clients m
LEFT JOIN known_clients k
    ON m.client_hash_id = k.client_hash_id
""").fetchdf()

display(client_join_audit)

print("Pairs by client-dimension match:")

display(con.execute("""
    SELECT
        CASE
            WHEN gsc_data_start IS NULL
                THEN 'Missing client dimension or GSC start date'
            ELSE 'Client dimension matched'
        END AS client_metadata_status,
        COUNT(*) AS client_content_pairs,
        COUNT(*) FILTER (
            WHERE impressions_march > 0
        ) AS pairs_with_impressions
    FROM march_dev
    GROUP BY 1
    ORDER BY 1
""").fetchdf())
```

```python
# April 2026 outcome aggregation — historical development only.
# June 2026 remains untouched.

april_path = (
    "hf://datasets/FlyRank/internship-warehouse/"
    "fact_content_daily_performance/month=2026-04/data_0.parquet"
)

con.execute(f"""
CREATE OR REPLACE TABLE april_outcomes AS
SELECT
    client_hash_id,
    content_hash_id,

    COUNT(DISTINCT report_date)
        FILTER (WHERE gsc_data_available IS TRUE)
        AS gsc_days_available_april,

    SUM(gsc_impressions)
        FILTER (WHERE gsc_data_available IS TRUE)
        AS impressions_april,

    SUM(gsc_clicks)
        FILTER (WHERE gsc_data_available IS TRUE)
        AS clicks_april,

    SUM(gsc_sum_position)
        FILTER (
            WHERE gsc_data_available IS TRUE
              AND gsc_avg_position IS NOT NULL
        ) AS position_sum_april,

    SUM(gsc_impressions)
        FILTER (
            WHERE gsc_data_available IS TRUE
              AND gsc_avg_position IS NOT NULL
        ) AS impressions_with_position_april

FROM read_parquet('{april_path}')
WHERE report_date >= DATE '2026-04-01'
  AND report_date < DATE '2026-05-01'
GROUP BY client_hash_id, content_hash_id
""")

con.execute("""
CREATE OR REPLACE TABLE april_outcomes_scored AS
SELECT
    *,
    100.0 * clicks_april
        / NULLIF(impressions_april, 0) AS ctr_april_pct,

    position_sum_april::DOUBLE
        / NULLIF(impressions_with_position_april, 0)
        AS weighted_position_april

FROM april_outcomes
""")

print("April outcome table created.")

print("\nApril coverage summary:")
display(con.execute("""
    SELECT
        COUNT(*) AS client_content_pairs,
        COUNT(*) FILTER (
            WHERE impressions_april > 0
        ) AS pairs_with_impressions,
        COUNT(*) FILTER (
            WHERE impressions_april >= 100
        ) AS pairs_with_at_least_100_impressions,
        COUNT(*) FILTER (
            WHERE impressions_april >= 300
        ) AS pairs_with_at_least_300_impressions,
        COUNT(*) FILTER (
            WHERE ctr_april_pct IS NOT NULL
        ) AS pairs_with_defined_ctr,
        COUNT(*) FILTER (
            WHERE weighted_position_april IS NOT NULL
        ) AS pairs_with_position
    FROM april_outcomes_scored
""").fetchdf())

print("\nApril impression distribution:")
display(con.execute("""
    SELECT
        COUNT(*) AS n_with_positive_impressions,
        ROUND(QUANTILE_CONT(impressions_april, 0.25), 1) AS p25,
        ROUND(QUANTILE_CONT(impressions_april, 0.50), 1) AS median,
        ROUND(QUANTILE_CONT(impressions_april, 0.75), 1) AS p75,
        ROUND(QUANTILE_CONT(impressions_april, 0.90), 1) AS p90,
        ROUND(QUANTILE_CONT(impressions_april, 0.95), 1) AS p95
    FROM april_outcomes_scored
    WHERE impressions_april > 0
""").fetchdf())

print("\nApril CTR distribution among pairs with at least 100 impressions:")
display(con.execute("""
    SELECT
        COUNT(*) AS n,
        ROUND(AVG(ctr_april_pct), 4) AS mean_ctr_pct,
        ROUND(MEDIAN(ctr_april_pct), 4) AS median_ctr_pct,
        ROUND(QUANTILE_CONT(ctr_april_pct, 0.25), 4) AS p25_ctr_pct,
        ROUND(QUANTILE_CONT(ctr_april_pct, 0.75), 4) AS p75_ctr_pct
    FROM april_outcomes_scored
    WHERE impressions_april >= 100
      AND ctr_april_pct IS NOT NULL
""").fetchdf())
```

```python
position_tier_audit = con.execute("""
WITH tiered AS (
    SELECT
        CASE
            WHEN weighted_position_april <= 3 THEN 'top_3'
            WHEN weighted_position_april <= 10 THEN 'positions_4_10'
            WHEN weighted_position_april <= 20 THEN 'positions_11_20'
            WHEN weighted_position_april <= 50 THEN 'positions_21_50'
            ELSE 'positions_51_plus'
        END AS position_tier,
        impressions_april,
        ctr_april_pct
    FROM april_outcomes_scored
    WHERE impressions_april > 0
      AND weighted_position_april IS NOT NULL
      AND ctr_april_pct IS NOT NULL
),
thresholds AS (
    SELECT 100 AS impression_floor
    UNION ALL
    SELECT 300 AS impression_floor
)
SELECT
    t.impression_floor,
    a.position_tier,
    COUNT(*) AS n_pages,
    ROUND(100.0 * AVG(
        CASE WHEN a.ctr_april_pct = 0 THEN 1.0 ELSE 0.0 END
    ), 2) AS zero_ctr_pct,
    ROUND(QUANTILE_CONT(a.ctr_april_pct, 0.25), 4) AS ctr_p25_pct,
    ROUND(MEDIAN(a.ctr_april_pct), 4) AS ctr_median_pct,
    ROUND(QUANTILE_CONT(a.ctr_april_pct, 0.75), 4) AS ctr_p75_pct
FROM tiered a
CROSS JOIN thresholds t
WHERE a.impressions_april >= t.impression_floor
GROUP BY t.impression_floor, a.position_tier
ORDER BY t.impression_floor, a.position_tier
""").fetchdf()

display(position_tier_audit)
```

```python
cohort_size_audit = con.execute("""
SELECT
    COUNT(*) FILTER (
        WHERE impressions_april >= 100
          AND weighted_position_april IS NOT NULL
    ) AS eligible_at_100,

    COUNT(*) FILTER (
        WHERE impressions_april >= 300
          AND weighted_position_april IS NOT NULL
    ) AS eligible_at_300,

    COUNT(DISTINCT client_hash_id) FILTER (
        WHERE impressions_april >= 100
          AND weighted_position_april IS NOT NULL
    ) AS clients_at_100,

    COUNT(DISTINCT client_hash_id) FILTER (
        WHERE impressions_april >= 300
          AND weighted_position_april IS NOT NULL
    ) AS clients_at_300
FROM april_outcomes_scored
""").fetchdf()

display(cohort_size_audit)
```

```python
# Join historical March features to April outcomes.
# Uses March position for tier assignment; April position is not used.

con.execute("""
CREATE OR REPLACE TABLE march_april_cohort AS
SELECT
    m.client_hash_id,
    m.content_hash_id,

    m.impressions_march,
    m.clicks_march,
    m.ctr_march_pct,
    m.weighted_position_march,
    m.gsc_days_available,
    m.content_type,
    m.search_volume,
    m.word_count,

    a.impressions_april,
    a.clicks_april,
    a.ctr_april_pct,

    CASE
        WHEN m.weighted_position_march <= 3 THEN 'top_3'
        WHEN m.weighted_position_march <= 10 THEN 'positions_4_10'
        WHEN m.weighted_position_march <= 20 THEN 'positions_11_20'
        WHEN m.weighted_position_march <= 50 THEN 'positions_21_50'
        ELSE 'positions_51_plus'
    END AS march_position_tier

FROM march_dev m
INNER JOIN april_outcomes_scored a
    ON m.client_hash_id = a.client_hash_id
   AND m.content_hash_id = a.content_hash_id

WHERE m.impressions_march > 0
  AND m.ctr_march_pct IS NOT NULL
  AND m.weighted_position_march IS NOT NULL
  AND a.impressions_april >= 300
  AND a.ctr_april_pct IS NOT NULL
""")

print("March-to-April cohort size:")
display(con.execute("""
    SELECT
        COUNT(*) AS eligible_pairs,
        COUNT(DISTINCT client_hash_id) AS clients,
        COUNT(DISTINCT content_hash_id) AS content_items,
        COUNT(*) FILTER (
            WHERE impressions_march >= 100
        ) AS pairs_with_100_plus_march_impressions,
        COUNT(*) FILTER (
            WHERE impressions_march >= 300
        ) AS pairs_with_300_plus_march_impressions
    FROM march_april_cohort
""").fetchdf())

print("Cohort by March position tier:")
display(con.execute("""
    SELECT
        march_position_tier,
        COUNT(*) AS n_pairs,
        ROUND(AVG(ctr_march_pct), 4) AS mean_march_ctr_pct,
        ROUND(MEDIAN(ctr_march_pct), 4) AS median_march_ctr_pct,
        ROUND(AVG(ctr_april_pct), 4) AS mean_april_ctr_pct,
        ROUND(MEDIAN(ctr_april_pct), 4) AS median_april_ctr_pct,
        ROUND(100.0 * AVG(
            CASE WHEN ctr_april_pct = 0 THEN 1.0 ELSE 0.0 END
        ), 2) AS april_zero_ctr_pct
    FROM march_april_cohort
    GROUP BY march_position_tier
    ORDER BY
        CASE march_position_tier
            WHEN 'top_3' THEN 1
            WHEN 'positions_4_10' THEN 2
            WHEN 'positions_11_20' THEN 3
            WHEN 'positions_21_50' THEN 4
            ELSE 5
        END
""").fetchdf())
```

```python
con.execute("""
CREATE OR REPLACE TABLE april_tier_reference AS
SELECT
    march_position_tier,
    COUNT(*) AS n_pairs,
    MEDIAN(ctr_april_pct) AS tier_median_ctr_april_pct,
    QUANTILE_CONT(ctr_april_pct, 0.25) AS tier_p25_ctr_april_pct,
    QUANTILE_CONT(ctr_april_pct, 0.75) AS tier_p75_ctr_april_pct
FROM march_april_cohort
GROUP BY march_position_tier
""")

con.execute("""
CREATE OR REPLACE TABLE opportunity_gap_audit AS
SELECT
    c.*,
    r.tier_median_ctr_april_pct,
    GREATEST(
        0.0,
        r.tier_median_ctr_april_pct - c.ctr_april_pct
    ) AS opportunity_gap_pp
FROM march_april_cohort c
JOIN april_tier_reference r
    USING (march_position_tier)
""")

print("April reference CTR by March position tier:")
display(con.execute("""
    SELECT
        march_position_tier,
        n_pairs,
        ROUND(tier_p25_ctr_april_pct, 4) AS tier_p25_ctr_pct,
        ROUND(tier_median_ctr_april_pct, 4) AS tier_median_ctr_pct,
        ROUND(tier_p75_ctr_april_pct, 4) AS tier_p75_ctr_pct
    FROM april_tier_reference
    ORDER BY
        CASE march_position_tier
            WHEN 'top_3' THEN 1
            WHEN 'positions_4_10' THEN 2
            WHEN 'positions_11_20' THEN 3
            WHEN 'positions_21_50' THEN 4
            ELSE 5
        END
""").fetchdf())

print("\nOpportunity-gap summary:")
display(con.execute("""
    SELECT
        COUNT(*) AS n_pairs,
        ROUND(AVG(opportunity_gap_pp), 4) AS mean_gap_pp,
        ROUND(MEDIAN(opportunity_gap_pp), 4) AS median_gap_pp,
        ROUND(QUANTILE_CONT(opportunity_gap_pp, 0.75), 4) AS p75_gap_pp,
        ROUND(QUANTILE_CONT(opportunity_gap_pp, 0.90), 4) AS p90_gap_pp,
        COUNT(*) FILTER (
            WHERE opportunity_gap_pp > 0
        ) AS pairs_below_tier_median,
        ROUND(
            100.0 * COUNT(*) FILTER (
                WHERE opportunity_gap_pp > 0
            ) / COUNT(*),
            2
        ) AS pct_below_tier_median
    FROM opportunity_gap_audit
""").fetchdf())

print("\nGap distribution by tier:")
display(con.execute("""
    SELECT
        march_position_tier,
        COUNT(*) AS n_pairs,
        COUNT(*) FILTER (
            WHERE opportunity_gap_pp > 0
        ) AS positive_gap_pairs,
        ROUND(AVG(opportunity_gap_pp), 4) AS mean_gap_pp,
        ROUND(MEDIAN(opportunity_gap_pp), 4) AS median_gap_pp
    FROM opportunity_gap_audit
    GROUP BY march_position_tier
    ORDER BY
        CASE march_position_tier
            WHEN 'top_3' THEN 1
            WHEN 'positions_4_10' THEN 2
            WHEN 'positions_11_20' THEN 3
            WHEN 'positions_21_50' THEN 4
            ELSE 5
        END
""").fetchdf())
```

```python
# Correct eligibility: use March information, not future April volume.
# April outcomes are observed only where April GSC impressions exist.

con.execute("""
CREATE OR REPLACE TABLE march_april_cohort_v2 AS
SELECT
    m.client_hash_id,
    m.content_hash_id,
    m.impressions_march,
    m.clicks_march,
    m.ctr_march_pct,
    m.weighted_position_march,
    m.gsc_days_available,
    m.content_type,
    m.search_volume,
    m.word_count,
    a.impressions_april,
    a.clicks_april,
    a.ctr_april_pct,

    CASE
        WHEN m.weighted_position_march <= 3 THEN 'top_3'
        WHEN m.weighted_position_march <= 10 THEN 'positions_4_10'
        WHEN m.weighted_position_march <= 20 THEN 'positions_11_20'
        WHEN m.weighted_position_march <= 50 THEN 'positions_21_50'
        ELSE 'positions_51_plus'
    END AS march_position_tier

FROM march_dev m
INNER JOIN april_outcomes_scored a
    ON m.client_hash_id = a.client_hash_id
   AND m.content_hash_id = a.content_hash_id

WHERE m.impressions_march >= 300
  AND m.ctr_march_pct IS NOT NULL
  AND m.weighted_position_march IS NOT NULL
  AND a.impressions_april > 0
  AND a.ctr_april_pct IS NOT NULL
""")

print("Primary cohort: eligible from March; measurable April outcome")
display(con.execute("""
    SELECT
        COUNT(*) AS eligible_pairs,
        COUNT(DISTINCT client_hash_id) AS clients,
        COUNT(*) FILTER (
            WHERE impressions_april >= 300
        ) AS also_at_least_300_april_impressions,
        COUNT(*) FILTER (
            WHERE march_position_tier = 'positions_51_plus'
        ) AS positions_51_plus_pairs
    FROM march_april_cohort_v2
""").fetchdf())

print("March-eligible pairs without a measurable April outcome")
display(con.execute("""
    SELECT
        COUNT(*) AS march_eligible_pairs,
        COUNT(*) FILTER (
            WHERE a.content_hash_id IS NULL
        ) AS no_april_fact_record,
        COUNT(*) FILTER (
            WHERE a.content_hash_id IS NOT NULL
              AND COALESCE(a.impressions_april, 0) = 0
        ) AS april_record_without_impressions
    FROM march_dev m
    LEFT JOIN april_outcomes_scored a
        ON m.client_hash_id = a.client_hash_id
       AND m.content_hash_id = a.content_hash_id
    WHERE m.impressions_march >= 300
      AND m.ctr_march_pct IS NOT NULL
      AND m.weighted_position_march IS NOT NULL
""").fetchdf())
```

```python
import numpy as np
import pandas as pd

# Reproducible split of distinct clients, not individual page rows.
client_ids = con.execute("""
    SELECT DISTINCT client_hash_id
    FROM march_april_cohort_v2
    ORDER BY client_hash_id
""").fetchdf()["client_hash_id"].tolist()

rng = np.random.RandomState(42)
shuffled_clients = rng.permutation(client_ids)

n_test_clients = max(1, round(len(shuffled_clients) * 0.20))
test_clients = set(shuffled_clients[:n_test_clients])
train_clients = set(shuffled_clients[n_test_clients:])

split_df = pd.DataFrame({
    "client_hash_id": client_ids,
    "split": [
        "test" if cid in test_clients else "train"
        for cid in client_ids
    ]
})

con.register("client_split_df", split_df)

con.execute("""
CREATE OR REPLACE TABLE client_split AS
SELECT * FROM client_split_df
""")

con.execute("""
CREATE OR REPLACE TABLE model_cohort AS
SELECT c.*, s.split
FROM march_april_cohort_v2 c
JOIN client_split s USING (client_hash_id)
WHERE c.march_position_tier != 'positions_51_plus'
""")

print("Client-level split summary:")
display(con.execute("""
    SELECT
        split,
        COUNT(DISTINCT client_hash_id) AS clients,
        COUNT(*) AS page_pairs
    FROM model_cohort
    GROUP BY split
    ORDER BY split
""").fetchdf())

print("Position tiers by split:")
display(con.execute("""
    SELECT
        split,
        march_position_tier,
        COUNT(*) AS page_pairs
    FROM model_cohort
    GROUP BY split, march_position_tier
    ORDER BY split, march_position_tier
""").fetchdf())
```

```python
con.execute("""
CREATE OR REPLACE TABLE train_tier_reference AS
SELECT
    march_position_tier,
    COUNT(*) AS train_pairs,
    MEDIAN(ctr_april_pct) AS train_median_april_ctr_pct
FROM model_cohort
WHERE split = 'train'
GROUP BY march_position_tier
""")

print("Training-only April CTR reference:")
display(con.execute("""
    SELECT
        march_position_tier,
        train_pairs,
        ROUND(train_median_april_ctr_pct, 4)
            AS train_median_april_ctr_pct
    FROM train_tier_reference
    ORDER BY march_position_tier
""").fetchdf())

con.execute("""
CREATE OR REPLACE TABLE labeled_cohort AS
SELECT
    c.*,
    r.train_median_april_ctr_pct,

    CASE
        WHEN c.ctr_april_pct < r.train_median_april_ctr_pct
            THEN 1
        ELSE 0
    END AS below_reference_label,

    GREATEST(
        0.0,
        r.train_median_april_ctr_pct - c.ctr_april_pct
    ) AS opportunity_gap_pp

FROM model_cohort c
JOIN train_tier_reference r USING (march_position_tier)
""")

print("Label balance by split:")
display(con.execute("""
    SELECT
        split,
        COUNT(*) AS page_pairs,
        SUM(below_reference_label) AS positive_labels,
        ROUND(AVG(below_reference_label) * 100, 2)
            AS positive_label_pct,
        ROUND(AVG(opportunity_gap_pp), 4)
            AS mean_opportunity_gap_pp
    FROM labeled_cohort
    GROUP BY split
    ORDER BY split
""").fetchdf())
```

```python
con.execute("""
CREATE OR REPLACE TABLE train_march_tier_reference AS
SELECT
    march_position_tier,
    MEDIAN(ctr_march_pct) AS train_median_march_ctr_pct
FROM model_cohort
WHERE split = 'train'
GROUP BY march_position_tier
""")

con.execute("""
CREATE OR REPLACE TABLE ranked_cohort AS
SELECT
    c.*,
    r.train_median_march_ctr_pct,
    GREATEST(
        0.0,
        r.train_median_march_ctr_pct - c.ctr_march_pct
    ) AS baseline_opportunity_score
FROM labeled_cohort c
JOIN train_march_tier_reference r
    USING (march_position_tier)
""")

baseline_summary = con.execute("""
SELECT
    split,
    COUNT(*) AS page_pairs,
    ROUND(AVG(baseline_opportunity_score), 4)
        AS mean_baseline_score,
    ROUND(
        100.0 * AVG(
            CASE WHEN baseline_opportunity_score > 0 THEN 1.0 ELSE 0.0 END
        ),
        2
    ) AS pct_with_positive_baseline_score
FROM ranked_cohort
GROUP BY split
ORDER BY split
""").fetchdf()

display(baseline_summary)
```

```python
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier

model_df = con.execute(f"""
SELECT
    r.client_hash_id,
    r.content_hash_id,
    r.split,
    r.march_position_tier,
    r.impressions_march,
    r.ctr_march_pct,
    r.weighted_position_march,
    r.gsc_days_available,
    r.search_volume,
    r.word_count,
    r.content_type,
    c.main_intent,
    r.below_reference_label,
    r.opportunity_gap_pp,
    r.baseline_opportunity_score
FROM ranked_cohort r
LEFT JOIN read_parquet('{content_path}') c
    ON r.client_hash_id = c.client_hash_id
   AND r.content_hash_id = c.content_hash_id
""").fetchdf()

numeric_features = [
    "log_impressions_march",
    "ctr_march_pct",
    "weighted_position_march",
    "gsc_days_available",
    "search_volume",
    "word_count",
]

categorical_features = [
    "march_position_tier",
    "content_type",
    "main_intent",
]

model_df["log_impressions_march"] = np.log1p(
    model_df["impressions_march"].clip(lower=0)
)

feature_columns = numeric_features + categorical_features

train_df = model_df[model_df["split"] == "train"].copy()
test_df = model_df[model_df["split"] == "test"].copy()

X_train = train_df[feature_columns]
y_train = train_df["below_reference_label"].astype(int)

X_test = test_df[feature_columns]
y_test = test_df["below_reference_label"].astype(int)

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features),
])

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        min_samples_leaf=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1,
    )),
])

model.fit(X_train, y_train)
test_df["model_probability"] = model.predict_proba(X_test)[:, 1]

print("Training rows:", len(train_df))
print("Test rows:", len(test_df))
print("Features:", feature_columns)
print("Test positive-label rate:", round(100 * y_test.mean(), 2), "%")
print("Model trained successfully.")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, average_precision_score

# Use the same held-out pages and the same target for both rankings.
evaluation_df = test_df[
    [
        "below_reference_label",
        "opportunity_gap_pp",
        "baseline_opportunity_score",
        "model_probability",
    ]
].copy()

evaluation_df["below_reference_label"] = (
    evaluation_df["below_reference_label"].astype(int)
)

base_rate = evaluation_df["below_reference_label"].mean()
k = max(1, int(np.ceil(len(evaluation_df) * 0.10)))

def evaluate_ranking(df, score_column, ranking_name):
    ranked = df.sort_values(score_column, ascending=False)
    top_k = ranked.head(k)

    precision_at_k = top_k["below_reference_label"].mean()
    recall_at_k = (
        top_k["below_reference_label"].sum()
        / df["below_reference_label"].sum()
    )

    return {
        "ranking": ranking_name,
        "test_pages": len(df),
        "k_top_10pct": k,
        "base_rate_pct": round(100 * base_rate, 2),
        "precision_at_k_pct": round(100 * precision_at_k, 2),
        "lift_vs_base_rate": round(
            precision_at_k / base_rate, 3
        ) if base_rate > 0 else np.nan,
        "recall_at_k_pct": round(100 * recall_at_k, 2),
        "mean_gap_top_k_pp": round(
            top_k["opportunity_gap_pp"].mean(), 4
        ),
    }

ranking_results = pd.DataFrame([
    evaluate_ranking(
        evaluation_df,
        "baseline_opportunity_score",
        "March CTR baseline",
    ),
    evaluate_ranking(
        evaluation_df,
        "model_probability",
        "Random Forest",
    ),
])

display(ranking_results)

print("ML ROC-AUC:", round(
    roc_auc_score(
        evaluation_df["below_reference_label"],
        evaluation_df["model_probability"],
    ), 4
))

print("ML average precision:", round(
    average_precision_score(
        evaluation_df["below_reference_label"],
        evaluation_df["model_probability"],
    ), 4
))
```

```python
con.execute(f"""
CREATE OR REPLACE TABLE may_features AS
WITH monthly AS (
    SELECT
        client_hash_id,
        content_hash_id,

        COUNT(DISTINCT report_date)
            FILTER (WHERE gsc_data_available IS TRUE)
            AS gsc_days_available,

        SUM(gsc_impressions)
            FILTER (WHERE gsc_data_available IS TRUE)
            AS impressions_may,

        SUM(gsc_clicks)
            FILTER (WHERE gsc_data_available IS TRUE)
            AS clicks_may,

        SUM(gsc_sum_position)
            FILTER (
                WHERE gsc_data_available IS TRUE
                  AND gsc_avg_position IS NOT NULL
            )
            AS position_sum_may,

        SUM(gsc_impressions)
            FILTER (
                WHERE gsc_data_available IS TRUE
                  AND gsc_avg_position IS NOT NULL
            )
            AS impressions_with_position_may

    FROM read_parquet('{may_path}')
    WHERE report_date >= DATE '2026-05-01'
      AND report_date < DATE '2026-06-01'
    GROUP BY client_hash_id, content_hash_id
)
SELECT
    m.*,

    100.0 * clicks_may
        / NULLIF(impressions_may, 0)
        AS ctr_may_pct,

    position_sum_may::DOUBLE
        / NULLIF(impressions_with_position_may, 0)
        AS weighted_position_may,

    c.content_type,
    c.search_volume,
    c.word_count,
    c.main_intent,

    CASE
        WHEN position_sum_may IS NULL
          OR impressions_with_position_may = 0
            THEN NULL
        WHEN position_sum_may::DOUBLE
             / NULLIF(impressions_with_position_may, 0) <= 3
            THEN 'top_3'
        WHEN position_sum_may::DOUBLE
             / NULLIF(impressions_with_position_may, 0) <= 10
            THEN 'positions_4_10'
        WHEN position_sum_may::DOUBLE
             / NULLIF(impressions_with_position_may, 0) <= 20
            THEN 'positions_11_20'
        WHEN position_sum_may::DOUBLE
             / NULLIF(impressions_with_position_may, 0) <= 50
            THEN 'positions_21_50'
        ELSE 'positions_51_plus'
    END AS may_position_tier

FROM monthly m
LEFT JOIN read_parquet('{content_path}') c
    ON m.client_hash_id = c.client_hash_id
   AND m.content_hash_id = c.content_hash_id
""")

may_summary = con.execute("""
SELECT
    COUNT(*) AS page_pairs,
    COUNT(DISTINCT client_hash_id) AS clients,
    COUNT(*) FILTER (WHERE impressions_may > 0)
        AS pairs_with_impressions,
    COUNT(*) FILTER (WHERE ctr_may_pct IS NOT NULL)
        AS pairs_with_defined_ctr,
    COUNT(*) FILTER (WHERE weighted_position_may IS NOT NULL)
        AS pairs_with_position,
    COUNT(*) FILTER (WHERE content_type IS NULL)
        AS missing_content_type,
    COUNT(*) FILTER (WHERE main_intent IS NULL)
        AS missing_main_intent
FROM may_features
""").fetchdf()

display(may_summary)

display(con.execute("""
SELECT
    may_position_tier,
    COUNT(*) AS page_pairs
FROM may_features
WHERE impressions_may > 0
GROUP BY may_position_tier
ORDER BY may_position_tier
""").fetchdf())
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score
import numpy as np
import pandas as pd

# June is accessed only now, after the development protocol is frozen.
june_path = (
    "hf://datasets/FlyRank/internship-warehouse/"
    "fact_content_daily_performance/month=2026-06/data_0.parquet"
)

# Aggregate June outcomes, respecting GSC availability.
con.execute(f"""
CREATE OR REPLACE TABLE june_outcomes AS
SELECT
    client_hash_id,
    content_hash_id,
    SUM(gsc_impressions)
        FILTER (WHERE gsc_data_available IS TRUE) AS impressions_june,
    SUM(gsc_clicks)
        FILTER (WHERE gsc_data_available IS TRUE) AS clicks_june
FROM read_parquet('{june_path}')
WHERE report_date >= DATE '2026-06-01'
  AND report_date < DATE '2026-07-01'
GROUP BY client_hash_id, content_hash_id
""")

con.execute("""
CREATE OR REPLACE TABLE june_outcomes_scored AS
SELECT
    *,
    100.0 * clicks_june / NULLIF(impressions_june, 0)
        AS ctr_june_pct
FROM june_outcomes
""")

june_outcome_summary = con.execute("""
SELECT
    COUNT(*) AS june_page_pairs,
    COUNT(DISTINCT client_hash_id) AS june_clients,
    COUNT(*) FILTER (WHERE impressions_june > 0)
        AS pairs_with_june_impressions
FROM june_outcomes_scored
""").fetchdf()

display(june_outcome_summary)

# Join the June outcomes to the previously frozen May scoring cohort.
june_df = con.execute("""
SELECT
    m.client_hash_id,
    m.content_hash_id,
    m.may_position_tier,
    m.model_probability,
    m.baseline_opportunity_score,
    j.impressions_june,
    j.clicks_june,
    j.ctr_june_pct
FROM final_may_cohort m
JOIN june_outcomes_scored j
  ON m.client_hash_id = j.client_hash_id
 AND m.content_hash_id = j.content_hash_id
WHERE j.impressions_june > 0
  AND j.ctr_june_pct IS NOT NULL
""").fetchdf()

# Apply the frozen April reference medians; do not calculate new June medians.
june_df = june_df.merge(
    con.execute("""
        SELECT
            march_position_tier AS may_position_tier,
            train_median_april_ctr_pct
        FROM train_tier_reference
    """).fetchdf(),
    on="may_position_tier",
    how="left",
    validate="many_to_one",
)

if june_df["train_median_april_ctr_pct"].isna().any():
    raise ValueError("Missing frozen reference threshold for a position tier.")

june_df["june_below_reference_label"] = (
    june_df["ctr_june_pct"] < june_df["train_median_april_ctr_pct"]
).astype(int)

june_df["june_opportunity_gap_pp"] = np.maximum(
    0.0,
    june_df["train_median_april_ctr_pct"] - june_df["ctr_june_pct"],
)

base_rate = june_df["june_below_reference_label"].mean()
k = max(1, int(np.ceil(len(june_df) * 0.10)))

def evaluate_june_ranking(df, score_column, ranking_name):
    ranked = df.sort_values(score_column, ascending=False)
    top_k = ranked.head(k)

    precision = top_k["june_below_reference_label"].mean()
    total_positives = df["june_below_reference_label"].sum()
    recall = (
        top_k["june_below_reference_label"].sum() / total_positives
        if total_positives else np.nan
    )

    return {
        "ranking": ranking_name,
        "evaluated_pairs": len(df),
        "top_k": k,
        "base_rate_pct": round(100 * base_rate, 2),
        "precision_at_10_pct": round(100 * precision, 2),
        "lift_vs_base_rate": round(precision / base_rate, 3)
            if base_rate else np.nan,
        "recall_at_10_pct": round(100 * recall, 2)
            if not np.isnan(recall) else np.nan,
        "mean_june_gap_top_k_pp": round(
            top_k["june_opportunity_gap_pp"].mean(), 4
        ),
    }

june_ranking_results = pd.DataFrame([
    evaluate_june_ranking(
        june_df, "baseline_opportunity_score", "May CTR baseline"
    ),
    evaluate_june_ranking(
        june_df, "model_probability", "Frozen Random Forest"
    ),
])

print("May eligible rows:", len(final_may_cohort))
print("Rows with measurable June CTR:", len(june_df))
print("Held-out clients evaluated:", june_df["client_hash_id"].nunique())
print("June positive-label rate:", round(100 * base_rate, 2), "%")

display(june_ranking_results)

if june_df["june_below_reference_label"].nunique() == 2:
    print(
        "June ROC-AUC:",
        round(roc_auc_score(
            june_df["june_below_reference_label"],
            june_df["model_probability"],
        ), 4),
    )
    print(
        "June average precision:",
        round(average_precision_score(
            june_df["june_below_reference_label"],
            june_df["model_probability"],
        ), 4),
    )
else:
    print("June ROC-AUC and average precision are undefined: only one label class.")
```

```python
tier_results = []

for tier, group in june_df.groupby("may_position_tier"):
    n = len(group)
    k_tier = max(1, int(np.ceil(n * 0.10)))
    tier_base_rate = group["june_below_reference_label"].mean()

    for score_column, ranking_name in [
        ("baseline_opportunity_score", "May CTR baseline"),
        ("model_probability", "Frozen Random Forest"),
    ]:
        top = group.sort_values(
            score_column, ascending=False
        ).head(k_tier)

        precision = top["june_below_reference_label"].mean()
        positive_count = group["june_below_reference_label"].sum()

        tier_results.append({
            "may_position_tier": tier,
            "ranking": ranking_name,
            "page_pairs": n,
            "top_10pct_count": k_tier,
            "base_rate_pct": round(100 * tier_base_rate, 2),
            "precision_at_10_pct": round(100 * precision, 2),
            "lift_vs_tier_base_rate": round(
                precision / tier_base_rate, 3
            ) if tier_base_rate > 0 else np.nan,
            "recall_at_10_pct": round(
                100 * top["june_below_reference_label"].sum()
                / positive_count, 2
            ) if positive_count > 0 else np.nan,
        })

tier_results_df = pd.DataFrame(tier_results)
display(tier_results_df.sort_values(
    ["may_position_tier", "ranking"]
).reset_index(drop=True))
```

```python
%cd /content/jsk-flyrank-ml-internship

from pathlib import Path

Path("docs").mkdir(exist_ok=True)

html = r"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<meta name="description" content="A research study evaluating how machine learning can prioritize SEO pages for human review when click-through rate underperforms.">
<title>Which Pages Should Be Reviewed First? — FlyRank Research</title>
<style>
:root {
  --ink:#202620; --muted:#606a63; --paper:#fbfaf6;
  --line:#dedfd6; --accent:#315c49; --soft:#edf2eb;
}
* { box-sizing:border-box; }
html { scroll-behavior:smooth; }
body {
  margin:0; background:var(--paper); color:var(--ink);
  font:17px/1.75 Georgia,"Times New Roman",serif;
}
header, main, footer { width:min(900px, calc(100% - 40px)); margin:auto; }
header { padding:64px 0 34px; border-bottom:1px solid var(--line); }
.kicker, nav, .meta, .eyebrow, .small, footer {
  font-family:Arial,Helvetica,sans-serif;
}
.kicker, .eyebrow {
  color:var(--accent); text-transform:uppercase;
  letter-spacing:.12em; font-size:.76rem; font-weight:700;
}
h1 { font-size:clamp(2.3rem,6vw,4.5rem); line-height:1.08; letter-spacing:-.04em; margin:18px 0; }
h2 { font-size:2rem; line-height:1.2; letter-spacing:-.025em; margin:0 0 18px; }
h3 { font-size:1.2rem; line-height:1.35; margin:28px 0 8px; }
.deck { font-size:1.25rem; color:var(--muted); max-width:720px; }
.meta { font-size:.85rem; color:var(--muted); margin-top:26px; }
nav { padding:18px 0; border-bottom:1px solid var(--line); font-size:.85rem; display:flex; flex-wrap:wrap; gap:8px 18px; }
a { color:var(--accent); text-underline-offset:3px; }
main section { padding:42px 0; border-bottom:1px solid var(--line); scroll-margin-top:20px; }
.lede { font-size:1.15rem; }
.callout { background:var(--soft); padding:20px 24px; border-left:4px solid var(--accent); margin:22px 0; }
.small { font-size:.86rem; color:var(--muted); line-height:1.6; }
table { width:100%; border-collapse:collapse; font-family:Arial,Helvetica,sans-serif; font-size:.9rem; margin:20px 0; }
th,td { text-align:left; padding:12px 10px; border-bottom:1px solid var(--line); vertical-align:top; }
th { font-size:.78rem; text-transform:uppercase; letter-spacing:.04em; color:var(--muted); }
.metric-grid { display:grid; grid-template-columns:repeat(3,minmax(0,1fr)); gap:12px; margin:22px 0; }
.metric { border:1px solid var(--line); padding:16px; min-width:0; }
.metric strong { display:block; font:700 clamp(1.3rem,3vw,2rem)/1.3 Arial,sans-serif; color:var(--accent); }
.metric span { display:block; font: .82rem/1.5 Arial,sans-serif; color:var(--muted); }
.chart { margin:22px 0; padding:18px; border:1px solid var(--line); }
.bar-row { display:grid; grid-template-columns:150px 1fr 58px; gap:10px; align-items:center; margin:12px 0; font: .85rem Arial,sans-serif; }
.track { height:14px; background:#e6e8e0; overflow:hidden; border-radius:2px; }
.fill { height:100%; background:var(--accent); }
.fill.baseline { background:#9ba69b; }
.caption { font:.8rem/1.5 Arial,sans-serif; color:var(--muted); margin-top:14px; }
li { padding:3px 0; }
footer { padding:30px 0 60px; font-size:.85rem; color:var(--muted); }
@media(max-width:600px) {
  header { padding-top:40px; }
  .metric-grid { grid-template-columns:1fr; }
  .bar-row { grid-template-columns:105px 1fr 50px; gap:7px; font-size:.78rem; }
  table { font-size:.78rem; }
  th,td { padding:9px 6px; }
}
</style>
</head>
<body>
<header>
  <div class="kicker">Applied Machine Learning · Research Report</div>
  <h1>Which pages should be reviewed first when CTR underperforms?</h1>
  <p class="deck">Evaluating a client-held-out machine-learning ranking approach for prioritizing SEO pages for human review.</p>
  <div class="meta">FlyRank ML Internship Capstone · Evaluation of development and temporal test results</div>
</header>

<nav aria-label="Paper contents">
  <a href="#abstract">Abstract</a><a href="#introduction">Introduction</a>
  <a href="#data">Data</a><a href="#method">Method</a>
  <a href="#results">Results</a><a href="#limitations">Limitations</a>
  <a href="#recommendations">Recommendations</a><a href="#reproducibility">Reproducibility</a>
  <a href="#acknowledgments">Acknowledgments</a>
</nav>

<main>
<section id="abstract">
  <div class="eyebrow">01 · Abstract</div>
  <h2>Abstract</h2>
  <p>This study investigates whether a machine-learning ranking model can help prioritize SEO pages for human review when click-through rate (CTR) underperforms a position-tier reference. The task uses March features to rank pages against an April-derived reference label, with a separate June outcome evaluation intended to test temporal performance. A Random Forest model is compared with a simple baseline ranking using the same evaluation split and precision-at-10, lift, recall, ranking metrics, and CTR shortfall. On the June temporal evaluation, the model achieved precision-at-10 of 82.55%, compared with 80.65% for the baseline, a gain of 1.90 percentage points. These results support further testing of ranking-assisted review, but do not establish causation, production readiness, or generalization to unseen clients beyond the evaluated data.</p>
</section>

<section id="introduction">
  <div class="eyebrow">02 · Introduction</div>
  <h2>Introduction and problem statement</h2>
  <p>SEO teams may have many pages to inspect but limited time for manual diagnosis. A useful prioritization system should help reviewers decide which pages to examine first, rather than claim to determine the cause of a page's performance.</p>
  <p>This project evaluates a ranking task: identify pages whose observed CTR is below a frozen reference median for their position tier. The model's intended use is to order candidates for review. It does not predict Google's algorithm, prove that a page needs a particular change, or establish that changing content will improve performance.</p>
  <div class="callout"><strong>Research question.</strong> Does a Random Forest ranking approach improve the quality of the first ten pages selected for review compared with a simple baseline, including on a later temporal evaluation?</div>
</section>

<section id="data">
  <div class="eyebrow">03 · Data</div>
  <h2>Data and target definition</h2>
  <p>The supplied project materials describe data spanning January 27, 2025, through June 30, 2026. The reported March feature partition contains approximately 9.84 million rows; the April and May partitions contain approximately 11.7 million rows each. The evaluated modelling datasets are smaller page-level feature/label tables, not the raw partition counts.</p>
  <h3>Target and evaluation windows</h3>
  <ul>
    <li><strong>Development evaluation:</strong> March features and an April outcome label, using a client-held-out split.</li>
    <li><strong>Temporal evaluation:</strong> March features and June outcomes, evaluated on eight held-out clients.</li>
    <li><strong>Positive label:</strong> observed CTR below the frozen April reference median for the relevant feature-month position tier.</li>
    <li><strong>Position tiers:</strong> 11–20, 21–50, 4–10, and top 3. Pages at positions 51 and beyond were excluded because the reference median was zero.</li>
  </ul>
  <p>The nine reported model inputs are log impressions, CTR percentage, weighted position, Search Console days available, search volume, word count, March position tier, content type, and main intent.</p>
  <p class="small">Data counts, label rules, and feature descriptions are based on the recorded project results and supplied project documentation. Raw partition sizes should not be interpreted as the number of independent training examples.</p>
</section>

<section id="method">
  <div class="eyebrow">04 · Methodology</div>
  <h2>Model and comparison design</h2>
  <p>The reported model is a Random Forest classifier with 200 trees, minimum leaf size 10, balanced class weights, and random seed 42. Numeric features use median imputation; categorical features use most-frequent imputation followed by one-hot encoding.</p>
  <h3>Baseline and metrics</h3>
  <p>The model is compared with a simple baseline ranking on the same respective evaluation set. Precision-at-10 (P@10) measures the fraction of the first ten ranked pages that have the positive label. Lift compares that precision with the positive-label prevalence; recall-at-10 measures the fraction of all positive examples captured in the first ten. The report also records ROC AUC, average precision, and mean positive CTR shortfall among the top ten.</p>
  <h3>Split strategy</h3>
  <p>The development evaluation reports 29 training clients and 8 test clients. The June temporal evaluation reports 21,548 page-level examples across eight held-out clients. A client-held-out split reduces direct overlap of clients between training and testing, while the later outcome window provides a temporal check.</p>
  <div class="callout"><strong>Interpretation rule.</strong> All reported model associations and feature importances are descriptive. They do not show that a feature causes CTR underperformance or that acting on a ranking will improve SEO outcomes.</div>
</section>

<section id="results">
  <div class="eyebrow">05 · Results</div>
  <h2>Results: baseline versus model</h2>
  <h3>Development evaluation</h3>
  <p>The recorded development set contains 61,874 training pairs and 11,338 test pairs. The positive-label rate in the development evaluation is 51.52%.</p>
  <table>
    <thead><tr><th>Metric</th><th>Baseline</th><th>Model</th></tr></thead>
    <tbody>
      <tr><td>Precision at 10</td><td>80.69%</td><td>85.19%</td></tr>
      <tr><td>Lift at 10</td><td>1.566</td><td>1.654</td></tr>
      <tr><td>Recall at 10</td><td>15.67%</td><td>16.54%</td></tr>
      <tr><td>Mean positive CTR shortfall in top 10</td><td>0.1195 percentage points</td><td>0.0887 percentage points</td></tr>
    </tbody>
  </table>

  <h3>June temporal evaluation</h3>
  <p>The June evaluation contains 21,548 examples, with a positive-label rate of 55.35%.</p>
  <div class="metric-grid">
    <div class="metric"><strong>82.55%</strong><span>Model precision at 10</span></div>
    <div class="metric"><strong>+1.90 pp</strong><span>Model P@10 gain over baseline</span></div>
    <div class="metric"><strong>0.7908</strong><span>Model ROC AUC</span></div>
  </div>
  <table>
    <thead><tr><th>Metric</th><th>Baseline</th><th>Model</th></tr></thead>
    <tbody>
      <tr><td>Precision at 10</td><td>80.65%</td><td>82.55%</td></tr>
      <tr><td>Lift at 10</td><td>1.457</td><td>1.491</td></tr>
      <tr><td>Recall at 10</td><td>14.57%</td><td>14.92%</td></tr>
      <tr><td>ROC AUC</td><td>Not reported</td><td>0.7908</td></tr>
      <tr><td>Average precision</td><td>Not reported</td><td>0.7784</td></tr>
      <tr><td>Mean positive CTR shortfall in top 10</td><td>0.1236 percentage points</td><td>0.0582 percentage points</td></tr>
    </tbody>
  </table>

  <div class="chart" role="img" aria-label="Bar chart comparing June precision at ten: baseline 80.65 percent, model 82.55 percent.">
    <strong>June precision at 10</strong>
    <div class="bar-row"><span>Baseline</span><div class="track"><div class="fill baseline" style="width:80.65%"></div></div><span>80.65%</span></div>
    <div class="bar-row"><span>Random Forest</span><div class="track"><div class="fill" style="width:82.55%"></div></div><span>82.55%</span></div>
    <p class="caption">Figure 1. Reported precision among the ten highest-ranked pages. Bars use a 0–100% scale.</p>
  </div>

  <p>The June result shows a modest improvement in top-ten precision: 1.90 percentage points over the baseline. The reported mean positive CTR shortfall among the top ten is also lower for the model. These findings are encouraging for triage, but the number of items in a top-ten metric is small and the result alone does not establish operational value.</p>

  <h3>Performance by position tier in June</h3>
  <table>
    <thead><tr><th>Position tier</th><th>Examples</th><th>Positive rate</th><th>Baseline P@10</th><th>Model P@10</th></tr></thead>
    <tbody>
      <tr><td>11–20</td><td>7,022</td><td>51.45%</td><td>76.24%</td><td>76.53%</td></tr>
      <tr><td>21–50</td><td>7,231</td><td>71.15%</td><td>86.05%</td><td>86.88%</td></tr>
      <tr><td>4–10</td><td>7,040</td><td>42.43%</td><td>81.53%</td><td>74.29%</td></tr>
      <tr><td>Top 3</td><td>255</td><td>71.37%</td><td>100.00%</td><td>80.77%</td></tr>
    </tbody>
  </table>
  <p>The aggregate result hides tier-level differences. The model is slightly better in the 11–20 and 21–50 tiers, but the reported baseline performs better in the 4–10 and top-3 tiers. The top-3 comparison requires particular caution because the sample contains only 255 examples and top-ten metrics can be sensitive to a small number of ranked items.</p>

  <h3>Feature importance</h3>
  <p>The recorded Random Forest feature-importance results assign the largest importance to March CTR (0.5746), followed by log impressions (0.1567), weighted position (0.0957), word count (0.0742), search volume (0.0331), and Search Console availability (0.0238). These are model-specific importance scores, not causal effects. In particular, a high importance for CTR may partly reflect the relationship between the input and the constructed outcome label.</p>
</section>

<section id="limitations">
  <div class="eyebrow">06 · Limitations</div>
  <h2>Limitations and honest framing</h2>
  <ul>
    <li><strong>Ranking is not intervention evidence.</strong> The study evaluates identification and ordering of labelled examples; it does not test whether changes to those pages improve outcomes.</li>
    <li><strong>Restricted label definition.</strong> The outcome depends on a frozen position-tier median. Excluding positions 51 and beyond limits what can be concluded about those pages.</li>
    <li><strong>Top-ten sensitivity.</strong> Precision-at-10 is based on a very small shortlist. Small ranking changes can move the metric materially.</li>
    <li><strong>Tier heterogeneity.</strong> The model does not outperform the baseline in every June position tier.</li>
    <li><strong>Generalization is uncertain.</strong> Eight held-out clients and one reported temporal evaluation do not establish performance across all clients, industries, or future periods.</li>
    <li><strong>Reproducibility needs verification.</strong> The repository contains a notebook and development record, but this report does not independently verify that a clean runtime can rebuild the sealed evaluation frame and reproduce every reported metric from committed code and inputs.</li>
    <li><strong>Importance is not causality.</strong> Feature-importance values should not be interpreted as evidence that modifying a feature will change CTR.</li>
  </ul>
</section>

<section id="recommendations">
  <div class="eyebrow">07 · Recommendations</div>
  <h2>Ranked recommendations</h2>
  <ol>
    <li><strong>Use the ranking as a reviewer-assistance shortlist, not an automated decision.</strong> Require a human to inspect the underlying page and context before recommending changes.</li>
    <li><strong>Run further time-based evaluations.</strong> Repeat the evaluation on additional later periods and held-out clients before relying on the observed improvement.</li>
    <li><strong>Compare approaches separately by position tier.</strong> The June results differ across tiers. A hybrid strategy that keeps the baseline for top-3 and 4–10 pages while testing the model for positions 11–50 is a hypothesis for a future experiment, not a validated recommendation.</li>
    <li><strong>Make the pipeline reproducible.</strong> Commit the exact evaluation-frame build procedure, metric computation, environment details, and aggregate outputs, then verify the results from a clean runtime.</li>
    <li><strong>Measure downstream usefulness.</strong> In a future controlled evaluation, measure reviewer agreement, time saved, actionability, and subsequent outcomes without implying causation from ranking metrics alone.</li>
  </ol>
</section>

<section id="reproducibility">
  <div class="eyebrow">08 · Reproducibility</div>
  <h2>Reproducibility and project materials</h2>
  <p>Project repository and notebook links are provided below. The notebook and development record are the starting points for inspecting the reported work.</p>
  <ul>
    <li><a href="https://github.com/ch-jsk/jsk-flyrank-ml-internship">GitHub repository</a></li>
    <li><a href="https://github.com/ch-jsk/jsk-flyrank-ml-internship/blob/main/work/notebooks/capstone.ipynb">Capstone notebook</a></li>
    <li><a href="https://github.com/ch-jsk/jsk-flyrank-ml-internship/blob/main/work/notebooks/development_record.json">Development record</a></li>
  </ul>
  <p class="small">These links identify project materials; they are not a claim that the entire experiment has already been reproduced from a clean environment. A complete reproducibility check should rebuild the evaluation frame, rerun the baseline and model on the same splits, recompute all metrics, and compare them with the committed aggregate results. Confirm that any required scripts and artifacts are committed before treating the workflow as reproducible.</p>
</section>

<section id="acknowledgments">
  <div class="eyebrow">09 · Acknowledgments</div>
  <h2>Acknowledgments and data credit</h2>
  <p>Built on the <a href="https://flyrank.ai" target="_blank" rel="noopener noreferrer">FlyRank ML Internship dataset</a>.</p>
  <p>This report describes an applied internship experiment. It avoids publishing client identities, private queries, or client-specific URLs. The findings are limited to the data, target definition, split strategy, and metrics described above.</p>
</section>
</main>

<footer>
  <p>FlyRank ML Internship Capstone · Research report</p>
  <p>Interpret the results as an evaluation of a page-review ranking workflow, not as a guarantee of SEO performance.</p>
</footer>
</body>
</html>
"""

Path("docs/index.html").write_text(html, encoding="utf-8")

print("Created docs/index.html")
print("File size:", Path("docs/index.html").stat().st_size, "bytes")
print("Title present:", "<title>Which pages should be reviewed first?" in html)
print("Sections:", html.count('<section id='))
```

```python
%cd /content/jsk-flyrank-ml-internship

from pathlib import Path
import re
import json
import shutil

html_path = Path("docs/index.html")
record_path = Path("work/notebooks/development_record.json")
backup_path = Path("docs/index.before_tier_fix.html")

html = html_path.read_text(encoding="utf-8")
record = json.loads(record_path.read_text(encoding="utf-8"))
shutil.copy2(html_path, backup_path)

# Build the table from the recorded final metrics.
tier_labels = {
    "positions_11_20": "11–20",
    "positions_21_50": "21–50",
    "positions_4_10": "4–10",
    "top_3": "Top 3",
}

tier_rows = record["final_test"]["tier_results"]
table_rows = []

for tier_key, label in tier_labels.items():
    rows = [r for r in tier_rows if r["may_position_tier"] == tier_key]
    baseline = next(r for r in rows if r["ranking"] == "May CTR baseline")
    model = next(r for r in rows if r["ranking"] == "Frozen Random Forest")

    table_rows.append(
        "<tr>"
        f"<td>{label}</td>"
        f"<td>{baseline['page_pairs']:,}</td>"
        f"<td>{baseline['base_rate_pct']:.2f}%</td>"
        f"<td>{baseline['precision_at_10_pct']:.2f}%</td>"
        f"<td>{model['precision_at_10_pct']:.2f}%</td>"
        "</tr>"
    )

new_table = """
<table>
  <thead>
    <tr>
      <th>May position tier</th>
      <th>Evaluated page pairs</th>
      <th>June positive rate</th>
      <th>Baseline precision at top 10%</th>
      <th>Model precision at top 10%</th>
    </tr>
  </thead>
  <tbody>
    """ + "\n    ".join(table_rows) + """
  </tbody>
</table>
"""

pattern = re.compile(
    r"<h[1-3][^>]*>\s*Performance by position tier in June\s*</h[1-3]>"
    r".*?(?=<h[1-3]\b|\Z)",
    re.I | re.S
)

replacement = """
<h2>Performance by position tier in June</h2>
<p>
The following results evaluate the frozen model on May features against June
outcomes. Precision at the top 10% is calculated separately within each May
position tier; it is not the overall June precision.
</p>
""" + new_table + """
<p>
The aggregate result hides tier-level differences. The model improves
precision in positions 11–20 (75.68% versus 72.12%), positions 21–50
(86.46% versus 76.38%), and the top 3 (76.92% versus 73.08%). The May CTR
baseline performs better in positions 4–10 (81.11% versus 76.99%). The
top-3 tier contains only 255 evaluated page pairs, so its result should be
interpreted cautiously.
</p>
"""

updated, count = pattern.subn(replacement, html, count=1)

if count != 1:
    raise RuntimeError(
        "Expected to replace exactly one tier section; no changes were saved. "
        "Inspect the heading in docs/index.html."
    )

html_path.write_text(updated, encoding="utf-8")

print("Updated June tier section.")
print("Backup saved to:", backup_path)
print("HTML size:", html_path.stat().st_size, "bytes")
```

```python
%cd /content/jsk-flyrank-ml-internship

from pathlib import Path
from html.parser import HTMLParser
import json

html = Path("docs/index.html").read_text(encoding="utf-8")
record = json.loads(
    Path("work/notebooks/development_record.json").read_text(encoding="utf-8")
)
final = record["final_test"]

class PaperParser(HTMLParser):
    def __init__(self):
        super().__init__()
        self.errors = []
        self.tables = 0
        self.headings = 0
        self.title_text = False

    def handle_starttag(self, tag, attrs):
        if tag == "table":
            self.tables += 1
        if tag in ("h1", "h2", "h3"):
            self.headings += 1

parser = PaperParser()
parser.feed(html)
parser.close()

checks = {
    "Updated 11–20 baseline": "72.12%" in html,
    "Updated 11–20 model": "75.68%" in html,
    "Updated 21–50 baseline": "76.38%" in html,
    "Updated 21–50 model": "86.46%" in html,
    "Updated 4–10 baseline": "81.11%" in html,
    "Updated 4–10 model": "76.99%" in html,
    "Updated top-3 baseline": "73.08%" in html,
    "Updated top-3 model": "76.92%" in html,
    "Overall June baseline": f'{final["baseline_precision_at_10_pct"]:.2f}%' in html,
    "Overall June model": f'{final["model_precision_at_10_pct"]:.2f}%' in html,
    "Original backup exists": Path("docs/index.before_tier_fix.html").exists(),
    "No placeholder in paper": "PASTE-YOUR-DEPLOYED-PAPER-URL-HERE" not in html,
}

for label, passed in checks.items():
    print(f"{'PASS' if passed else 'CHECK'} — {label}")

print(f"\nHTML parsed successfully; headings: {parser.headings}, tables: {parser.tables}")
print("Paper bytes:", len(html.encode("utf-8")))
print("All checks passed:", all(checks.values()))
```
